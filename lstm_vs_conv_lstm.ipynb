## Statistics

```python
from numpy import mean
from scipy.stats import shapiro, ttest_ind

from src.sota.kfold import SOTAFoldCrossValidation
from src.hpe_dnn.kfold import HpeDnnFoldCrossValidation
from src.rnn.kfold import RnnFoldCrossValidation
from src.conv_lstm.kfold import ConvLstmFoldCrossValidation
```

```python
sota = SOTAFoldCrossValidation.evaluation_instance("yolo11n-full-kf", "yolo11n-cls")
dnn = HpeDnnFoldCrossValidation.evaluation_instance("arch2-no-bal-no-aug-kf")
rnn = RnnFoldCrossValidation.evaluation_instance("arch5-medium-dense")
conv_lstm = ConvLstmFoldCrossValidation.evaluation_instance("arch1-small")

sota_test_arr = sota.get_test_accuracy_metrics()
dnn_test_arr = dnn.get_test_accuracy_metrics()
rnn_test_arr = rnn.get_test_accuracy_metrics()
conv_lstm_test_arr = conv_lstm.get_test_accuracy_metrics()
```

```python
print(f"ConvLstm test accuracies are normally distributed?: {shapiro(rnn_test_arr)}")
```

```python
# t test for normally distributed data
print(f"Ho: Both test results have indentical averages?: {ttest_ind(sota_test_arr, conv_lstm_test_arr, alternative="greater")}")
print(f"Average test accuracy of SOTA yolo11n-full models ({mean(sota_test_arr):.1%}) is " +
      f"significantly greater than that of ConvLstm arch1-small models ({mean(conv_lstm_test_arr):.1%})")
```

```python
# t test for normally distributed data
print(f"Ho: Both test results have indentical averages?: {ttest_ind(conv_lstm_test_arr, dnn_test_arr, alternative="greater")}")
print(f"Average test accuracy of ConvLstm arch1-small models ({mean(conv_lstm_test_arr):.1%}) is " +
      f"significantly greater than that of DNN arch2-no-bal-no-aug-full models ({mean(dnn_test_arr):.1%})")
```

```python
# t test for normally distributed data
print(f"Ho: Both test results have indentical averages?: {ttest_ind(rnn_test_arr, conv_lstm_test_arr, alternative="greater")}")
print(f"Average test accuracy of RNN arch5-medium-dense models ({mean(rnn_test_arr):.1%}) is " +
      f"significantly greater than that of ConvLstm arch1-small models ({mean(conv_lstm_test_arr):.1%})")
```
