# CONV LSTM

```python
from src.conv_lstm.architecture import ConvLstmArch
from src.conv_lstm.kfold import ConvLstmFoldCrossValidation
from src.conv_lstm.model import ConvLstmConstructorArgs, ConvLstmModelInitializeArgs, \
ConvLstmIntMultiRunTrainArgs, ConvLstmIntTrainArgs

from src.common.model import DEFAULT_DATASET, DEFAULT_FULL_DATASET
```

## Model Selection

```python
# arch1

conv_lstm_folds = ConvLstmFoldCrossValidation(
    model_args=ConvLstmConstructorArgs(
        name="arch1-small",
        model_initialize_args=ConvLstmModelInitializeArgs(
            model_arch=ConvLstmArch.ARCH1,
            input_width=5,
            spacing=1
        ),
        dataset_name=DEFAULT_FULL_DATASET
    )
)
conv_lstm_folds.train_folds(train_run_args=ConvLstmIntMultiRunTrainArgs(
    runs=5,
    train_args=ConvLstmIntTrainArgs(
        epochs=10
    )
))
```
